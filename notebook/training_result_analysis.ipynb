# Training Result Analysis Part

```python
from pathlib import Path
from IPython.display import Image as IPImage, display

print("YOLO26m Training Visualizations\n")
print("=" * 70)

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "runs" / "detect" / "val").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

RESULTS_DIR = PROJECT_ROOT / "runs" / "detect" / "val"
if not RESULTS_DIR.exists():
    RESULTS_DIR = PROJECT_ROOT / "runs" / "detection" / "train"

plot_categories = {
    'Performance Curves': [
        ('Overall Results', RESULTS_DIR / 'results.png'),
        ('Box F1 Score', RESULTS_DIR / 'BoxF1_curve.png'),
        ('Box Precision', RESULTS_DIR / 'BoxP_curve.png'),
        ('Box Recall', RESULTS_DIR / 'BoxR_curve.png'),
        ('Precision-Recall Curve', RESULTS_DIR / 'BoxPR_curve.png')
    ],
    'Confusion Matrices': [
        ('Raw Counts', RESULTS_DIR / 'confusion_matrix.png'),
        ('Normalized', RESULTS_DIR / 'confusion_matrix_normalized.png')
    ],
    'Training Examples': [
        ('Training Batch Sample', RESULTS_DIR / 'train_batch0.jpg'),
        ('Labels Distribution', RESULTS_DIR / 'labels.jpg')
    ],
    'Validation Predictions': [
        ('Ground Truth Labels', RESULTS_DIR / 'val_batch0_labels.jpg'),
        ('Model Predictions', RESULTS_DIR / 'val_batch0_pred.jpg')
    ]
}

for category, plots in plot_categories.items():
    print(f"\n{category}:")
    print("-" * 70)

    for title, filepath in plots:
        if filepath.exists():
            print(f"\n{title}:")
            display(IPImage(filename=str(filepath), width=900))
        else:
            print(f"\n{title}: Not generated")

print("\n" + "=" * 70)
print("Visualization Summary Complete")
```
